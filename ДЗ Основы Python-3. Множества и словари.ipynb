# Задание 1. Множества

```python
hw_set = {1,3,8,9,10,15,16,16}
```

```python
##############################################   Задание 1.1   #################################################################

### Как вы думаете, чему равна длина данного множества? Выведите его длину с помощью функции len

print(len(hw_set))
```

```python
##############################################   Задание 1.2   #################################################################

### Создайте пустое множество hw_set_sq и с помощью цикла запишите в него возведенные в квадрат элементы множества hw_set

hw_set_sq = set()
for item in hw_set:
    hw_set_sq.add(item ** 2)

print(hw_set_sq)
```

```python
##############################################   Задание 1.3   #################################################################

### Добавьте к hw_set элементы 'my_string' и 'your_string'
### Повторите действия из 1.2, но добавьте проверку на тип данных

hw_set_sq = set()
for item in hw_set:
    if isinstance(item, (int, float)): 
        hw_set_sq.add(item ** 2)
print(hw_set)
print(hw_set_sq)
```

```python
##############################################   Задание 1.4   #################################################################

### Создайте копию множества hw_set (назовите его hw_set_copy)
### Используйте цикл (счетчик пробегает по hw_set_copy) и условия для модификации множества hw_set
### Добавьте к множеству hw_set квадраты всех цифровых элементов и первые символы всех строк

hw_set = {1,3,8,9,10,15,16,16,'my_string','your_string'}
hw_set_copy = hw_set.copy()

for item in hw_set_copy: 
    if isinstance(item, str) and len(item) > 0:
        hw_set.add(item[0])
print(hw_set)
print(hw_set_copy)
```

```python
##############################################   Задание 1.5   #################################################################

### Создайте копию множества hw_set (назовите его hw_set_copy)
### Используйте цикл (счетчик пробегает по hw_set_copy) и условия для модификации множества hw_set
### Удалите из множества hw_set все элементы-строки

hw_set = {1,3,8,9,10,15,16,16,'my_string','your_string'}
hw_set_copy = hw_set.copy()

for item in hw_set_copy:
    if isinstance(item, str):
        hw_set.remove(item)

print(hw_set)
```

# Задание 2. Словари

```python
employees_0 = {'Bob':'Dev','Albert':'Manager','Victor':'Analyst','Anna':'Analyst'}
```

```python
##############################################   Задание 2.1   #################################################################

### Выведите имена всех сотрудников, кто не является менеджером. Используйте цикл for с условием внутри.

for name, job in employees_0.items():
    if job != 'Manager':
        print(f"   {name} ({job})")
```

```python
##############################################   Задание 2.2   #################################################################

### Посчитайте количество сотрудников-аналитиков. Используйте цикл for с условием внутри.

count_analysts = 0
for name, job in employees_0.items():
    if job == 'Analyst':
        count_analysts += 1
print(count_analysts)
```

```python
##############################################   Задание 2.3   #################################################################

### Посчитайте количество сотрудников, которые или являются аналитиками, или их зовут Боб.
### Используйте цикл for с условием внутри.

count_target = 0
for name, job in employees_0.items():
    if job == 'Analyst' or name == 'Bob':
        count_target += 1
print(count_target)
```

```python
##############################################   Задание 2.4   #################################################################

##### Изменим немного структуру словаря

employees_1 = {'name':['Bob','Albert','Victor', 'Anna'], 'job':['Dev','Manager','Analyst','Analyst']}

### Выведите профессию первого сотрудника в словаре

print(employees_1['job'][0])
```

```python
##############################################   Задание 2.5   #################################################################

### Выведите количество сотрудников-аналитиков в словаре. Используйте цикл for с условием внутри.

count_analysts_1 = 0
for i in range(len(employees_1['job'])):
    if employees_1['job'][i] == 'Analyst':
        count_analysts_1 += 1
print(count_analysts_1)
```

# Задание 3. Цикл while

```python
##############################################   Задание 3.1   #################################################################

### С помощью цикла while записывайте в список lst_sum все элементы с конца списка lst_init, пока сумма элементов в lst_sum не достигнет 100
### Вам потребуются методы списков append и pop

lst_init = [4, 17, 22, 10, 8, 15, 8, 9, 2, 32, 13, 11, 12, 10, 9, 12, 22, 23, 13, 11, 10, 7, 42, 43, 14, 15, 16]
lst_sum = []

while sum(lst_sum) < 100:
    last_elem = lst_init.pop() 
    lst_sum.append(last_elem) 

print(lst_sum)
print(sum(lst_sum))
print(lst_init)
```

```python
##############################################   Задание 3.2   #################################################################

### Создайте переменную bin_var, которая является параметром и может принимать значения 1 или 0.
### Если она равняется другому, то выводится сообщение "Переменная bin_var должна равняться или 1, или 0!"
### Если она равняется нулю, то выполняется действие из задания 3.1
### Если она равняется единице, то действие из задания 3.1 выполняется не с конца списка lst_init, а с начала
### Для этого вам потребуется метод списка reverse(). Не забудьте пересоздать развернутый наоборот список внутри условия

lst_init = [4, 17, 22, 10, 8, 15, 8, 9, 2, 32, 13, 11, 12, 10, 9, 12, 22, 23, 13, 11, 10, 7, 42, 43, 14, 15, 16]
lst_sum = []

bin_var = 0 

if bin_var == 1:
    temp_lst_init = lst_init.copy()
    temp_lst_init.reverse()
    while sum(lst_sum) < 100:
        lst_sum.append(temp_lst_init.pop())
    print("3.2 (bin_var=1) Действие с начала. lst_sum:", lst_sum)
    print("Сумма:", sum(lst_sum))

elif bin_var == 0:
    temp_lst_init = lst_init.copy()
    while sum(lst_sum) < 100:
        lst_sum.append(temp_lst_init.pop())
    print("3.2 (bin_var=0) Действие с конца. lst_sum:", lst_sum)
    print("Сумма:", sum(lst_sum))

else:
    print("Ошибка: bin_var должен быть 0 или 1")
```
