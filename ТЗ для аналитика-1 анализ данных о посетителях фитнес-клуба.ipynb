Вам предстоит поработать со списком возрастов посетителей фитнес-клубов age_lst

**Задача 1.** Рассчитайте средний возраст по всему списку age_lst

**Задача 2.** Рассчитайте средний возраст для посетителей, которые моложе 50 лет

**Задача 3.** Оберните предыдущий код в функцию, которая на вход принимает список и пороговое значение

**Задача 4.** Рассчитайте среднее значение для всех элементов списка, за исключением тех, которые лежат вне интервала [Q1-1,5*IQR; Q3+1,5*IQR]

**Задача 5 (со звёздочкой).** Оберните предыдущий код в функцию, которая принимает на вход список и рассчитывает средний возраст, исключая из расчётов выбросы

```python
age_lst = [23, 45, 18, 80, 22, 25, 34, 37, 41, 88, 19, 20, 18, 45, 41, 39, 31, 32, 22, 28, 37]
import statistics
```

```python
# Задача 1

avg_all = sum(age_lst) / len(age_lst)
print(f"Задача 1. Средний возраст: {avg_all:.2f}")
```

```python
# Задача 2

filtered = [a for a in age_lst if a < 50]
avg_filtered = sum(filtered) / len(filtered)
print(f"Задача 2. Средний возраст (<50): {avg_filtered:.2f}")
```

```python
# Задача 3

def avg_by_threshold(ages, threshold):
    filtered = [a for a in ages if a <= threshold]
    return sum(filtered) / len(filtered) if filtered else None

print(f"Задача 3. Средний возраст (порог 50): {avg_by_threshold(age_lst, 50):.2f}")
```

```python
# Задача 4

sorted_ages = sorted(age_lst)
q1 = statistics.quantiles(sorted_ages, n=4)[0]
q3 = statistics.quantiles(sorted_ages, n=4)[2]
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

clean_ages = [a for a in age_lst if lower <= a <= upper]
avg_clean = sum(clean_ages) / len(clean_ages)
print(f"Задача 4. Средний возраст без выбросов: {avg_clean:.2f}")
```

```python
# Задача 5

def avg_no_outliers(data):
    if len(data) < 4:
        return None
    s = sorted(data)
    q1 = statistics.quantiles(s, n=4)[0]
    q3 = statistics.quantiles(s, n=4)[2]
    iqr = q3 - q1
    low = q1 - 1.5 * iqr
    high = q3 + 1.5 * iqr
    filtered = [x for x in data if low <= x <= high]
    return sum(filtered) / len(filtered) if filtered else None

print(f"Задача 5. Средний возраст (функция): {avg_no_outliers(age_lst):.2f}")
```
