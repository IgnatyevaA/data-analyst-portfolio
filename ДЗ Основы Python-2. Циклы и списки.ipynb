### Задача 2.
Теперь у вас есть список платежей payments.

**Задание:**

+ Выведите каждый элемент списка payments, используя цикл for.

```python
payments = [1200, 3000, 120000, 10800, 3000, 18300]
```

### Задача 3
Используя список payments из предыдущей задачи, найдите средний платёж.

**Задание:**

Рассчитайте средний платёж двумя способами:

Вручную — с помощью цикла:

+ посчитайте сумму всех платежей

+ посчитайте количество платежей

+ найдите среднее как отношение суммы к количеству
(не используйте функции sum() и len())

С помощью встроенных функций — используя sum() и len()

Сравните результаты: они должны совпадать.

```python
# Решение
total = 0
count = 0
for i in payments:
    total += i
    count += 1
avg = total / count
print(total)
print(count)
print(avg)
```

```python
avg_builtin = sum(payments) / len(payments)
print(sum(payments))
print(len(payments))
print(avg_builtin)
```

### Задача 4. 

Менеджер хочет понять, какие платежи были крупными.

**Задание:**

Создайте новый список *big_payments*.

Добавьте в него только те значения из списка payments, которые больше 10000 рублей.

Выведите список *big_payments*.

Затем выведите индексы этих платежей в исходном списке payments вместе с их значениями в формате:
индекс: значение

**Подсказка:**

+ используйте цикл for

+ для получения индекса используйте range(len(payments))

```python
# Решение

big_payments = []

for i in payments:
    if i > 10000:
        big_payments.append(i)

print(big_payments)

for i in range(len(payments)):
    if payments[i] > 10000:
        print(i, ":", payments[i])
```

### Задача 5.

У вас есть список *clients_payments*, в котором хранятся платежи клиентов.

В списке могут встречаться:

+ отдельные числа (один платёж),

+ списки с несколькими платежами.

**Задание:**

Выведите все платежи из списка clients_payments, включая те, которые находятся внутри вложенных списков.
Каждое значение выведите на отдельной строке.

**Подсказка:**

+ используйте цикл for, чтобы перебрать элементы списка

+ если элемент — число, выведите его

+ если элемент — список, пройдитесь по нему с помощью вложенного цикла



```python
clients_payments = [300, 1000, [1200, 3000], 400, [560]]
```

```python
# Решение

for item in clients_payments:
    if isinstance(item, list):
        for payment in item:
            print(payment)
    else:
        print(item)
```

### Задача 6.

Теперь поработаем с более "грязными" данными. Вам прислали новый список **clients_payments**.

В списке могут встречаться:

+ строки (имена клиентов),

+ числа (отдельные платежи),

+ списки и кортежи с несколькими платежами.

**Задание:**

Выведите все числовые значения (платежи) из списка clients_payments, включая те, которые находятся внутри вложенных списков и кортежей.
Каждое значение выведите на отдельной строке.

**Подсказка:**

+ используйте цикл for, чтобы перебрать элементы списка

+ если элемент — число, выведите его

+ если элемент — список или кортеж, пройдитесь по нему с помощью вложенного цикла

+ игнорируйте все нечисловые значения

```python
clients_payments = [
    "Иван",
    1000,
    [1200, 300, 800],
    "Мария", 
    "ошибка ввода",
    (4000, 5000)
]
```

```python
# Решение

clear_list = []

for i in clients_payments:
    if (type(i)==list) | (type(i)==tuple):
        for j in i:
            clear_list.append(j)
    elif (type(i)==int):
        clear_list.append(i)
    else:
        pass

print(clear_list)
```
