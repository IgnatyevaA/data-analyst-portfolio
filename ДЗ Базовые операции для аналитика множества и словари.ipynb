**Задача 3**

У вас есть словарь stock с остатками товаров. Нужно узнать, сколько единиц есть в наличии для заданных товаров из списка check_list.
Если товара нет — вывести "Нет в наличии".

*Подсказки:*

+ Переберите товары из списка check_list с помощью for.

+ Для каждого используйте stock.get(товар, "Нет в наличии").

+ Выведите название товара и результат.

```python
stock = {'ноутбук': 15, 'смартфон': 8, 'планшет': 5, 'наушники': 0}
check_list = ['планшет', 'монитор', 'ноутбук']
```

```python
for item in check_list:
    quantity = stock.get(item)
    
    if quantity is not None:
        print(f"{item}: {quantity}")
    else:
        print(f"{item}: Нет в наличии")
```

**Задача 4**

Есть словарь с ценами товаров. Нужно вывести, какие товары стоят дороже 10 000 ₽,
а какие 10 000 или дешевле.

*Подсказка:*

+ создайте два пустых списка

+ Используйте метод .items() — он позволяет перебрать и ключ, и значение словаря.

+ Внутри цикла for сравнивайте цену (price) с 10 000 и добавляйте товар в соответствующий список с помощью .append().

```python
prices = {'ноутбук': 55000, 'смартфон': 22000, 'мышь': 1500, 'клавиатура': 3500, 'планшет': 18000}
```

```python
expensive = []
cheap = []

for item, price in prices.items():
    if price > 10000:
        expensive.append(item)
    else:
        cheap.append(item)

print(expensive)
print(cheap)
```

**Задача 5**

У вас есть словарь sales с продажами сотрудников по месяцам.
Нужно сформировать список сотрудников, которые в марте сделали продаж больше 15 000 ₽.

*Подсказка:*

+ Переберите сотрудников в цикле for.

+ Обращайтесь к значениям по двум ключам.

```python
sales = {
    'Анна': {'январь': 12000, 'февраль': 15000, 'март': 17000},
    'Борис': {'январь': 20000, 'февраль': 21000, 'март': 19000},
    'Виктория': {'январь': 14000, 'февраль': 13000, 'март': 16000},
    'Григорий': {'январь': 18000, 'февраль': 15000, 'март': 9000}
}
```

```python
top_sellers = []

for name, months in sales.items():
    march_sales = months['март']
    if march_sales > 15000:
        top_sellers.append(name)

print(top_sellers)
```

**Задача 6**

Интернет-магазин ставит цель — заработать 100 000 ₽ за месяц.
Известна выручка по дням (список sales). Нужно посчитать, через сколько дней сумма продаж превысит план.
Используйте while, чтобы накапливать выручку по одному дню.

*Подсказка*:

+ Создаем переменные для накопления и счетчик дней: заведите total = 0 для суммы продаж и day = 0 для подсчета дней.

+ Используем while для накопления: пока total меньше цели, добавляйте продажи текущего дня (sales[day]) и увеличивайте day на 1.

```python
sales = [12000, 15000, 8000, 20000, 17000, 25000, 10000]
target = 100000
```

```python
total = 0
day = 0

while total < target:
    total += sales[day]
    day += 1

print(f"План в {target} ₽ будет достигнут на {day}-й день.")
print(f"Накопленная сумма: {total} ₽")
```
