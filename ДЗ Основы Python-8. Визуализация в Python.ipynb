```python
##############################################   Задание 1.1   #################################################################

######## Постройте график с помощью plt.scatter, где по горизонтальной оси будут значения Х, а по вертикальной оси - значения Y.
######## Добавьте название графика "Мой первый график"
######## Добавьте подпись горизонтальной оси "Ряд Х"
######## Добавьте подпись вертикальной оси "Ряд Y"
######## Добавьте сетку на график с помощью функции grid


from matplotlib import pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7, 8, 9]
y = [8, 5, 6, 10, 18, 26, 18, 22, 23]

plt.scatter(x, y)
plt.title('Мой первый график')
plt.xlabel('Ряд Х')
plt.ylabel('Ряд Y')
plt.grid()
plt.show()
```

```python
##############################################   Задание 1.2   #################################################################

######## Постройте график с помощью plt.plot, где по горизонтальной оси будут значения Х, а по вертикальной оси - значения Y.
######## Повторите все те же самые действия по оформлению графика, как в прошлом пункте.
######## Установите размер графика 7*7 с помощью параметра figsize функции figure

plt.figure(figsize=(7, 7))
plt.plot(x, y)
plt.title('Мой первый график')
plt.xlabel('Ряд Х')
plt.ylabel('Ряд Y')
plt.grid()
plt.show()
```

```python
##############################################   Задание 1.3   #################################################################

##### Добавьте на визуализацию из прошлого пункта сплошную вертикальную линию красного цвета в точке 6 
##### Добавьте также пунктирную вертикальную линию красного цвета в точке 2
##### (чтобы сделать линию пунктирной, воспользуйтесь аргументом linestyle)
##### Сделайте кривую y(x) зеленого цвета
##### Добавьте легенду для максимума и минимума в правый нижний угол
##### Для этого загуглите как работает функция axvline

plt.figure(figsize=(7, 7))
plt.plot(x, y, 'g')
plt.axvline(6, color='red')
plt.axvline(2, color='red', linestyle='--')
plt.title('Мой первый график')
plt.xlabel('Ряд Х')
plt.ylabel('Ряд Y')
plt.grid()
plt.legend(['Максимум', 'Минимум'], loc='lower right')
plt.show()
```

```python
##############################################   Задание 1.4   #################################################################

##### Добавьте на график еще одную кривую z = [6,9,13,13,16,12,20,18,17] и сделайте ее фиолетовго цвета
##### Добавьте сплошную вертикальную линию черного цвета в точке 7
##### Добавьте пунктирную вертикальную линию черного цвета в точке 1
##### Подпишите их на легенде как Максимум Z и Минимум Z соответственно

z = [6, 9, 13, 13, 16, 12, 20, 18, 17]

plt.figure(figsize=(7, 7))
plt.plot(x, y, 'g', label='Ряд Y')
plt.plot(x, z, 'purple', label='Ряд Z')
plt.axvline(6, color='red', label='Максимум Y')
plt.axvline(2, color='red', linestyle='--', label='Минимум Y')
plt.axvline(7, color='black', label='Максимум Z')
plt.axvline(1, color='black', linestyle='--', label='Минимум Z')
plt.title('Мой первый график')
plt.xlabel('Ряд Х')
plt.ylabel('Ряд Y')
plt.grid()
plt.legend()
plt.show()
```

```python
##############################################   Задание 1.5   #################################################################

###### Воспроизведите тот же график, что и в прошлом пункте, используя ax и subplots

z = [6, 9, 13, 13, 16, 12, 20, 18, 17]

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(x, y, 'g', label='Ряд Y')
ax.plot(x, z, 'purple', label='Ряд Z')
ax.axvline(6, color='red', label='Максимум Y')
ax.axvline(2, color='red', linestyle='--', label='Минимум Y')
ax.axvline(7, color='black', label='Максимум Z')
ax.axvline(1, color='black', linestyle='--', label='Минимум Z')
ax.set_title('Мой первый график')
ax.set_xlabel('Ряд Х')
ax.set_ylabel('Ряд Y')
ax.grid()
ax.legend()
plt.show()
```

```python
##############################################   Задание 1.6   #################################################################

##### Создайте дополнительную ось Y справа (используйте twinx())
##### На этой оси постройте столбчатую диаграмму, отображающую разницу Y-Z
##### Сделайте столбцы полупрозрачными: alpha=0.3
##### Добавьте label='difference' для отображения в легенде
##### Расположите легенды:
#####   - Основную легенду — снизу
#####   - Легенду дополнительной оси — справа-сверху

z = [6, 9, 13, 13, 16, 12, 20, 18, 17]
diff = [y[i] - z[i] for i in range(len(y))]

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(x, y, 'g', label='Ряд Y')
ax.plot(x, z, 'purple', label='Ряд Z')

ax2 = ax.twinx()
ax2.bar(x, diff, alpha=0.3, label='difference')

ax.legend(loc='lower center')
ax2.legend(loc='upper right')
plt.show()
```

```python
##############################################   Задание 1.7   #################################################################

#####  Добавьте ограничение на правую ось для графика из предыдущего задания.
#####  Пусть на графике показываются только положительные значения разницы
#####  Переместите основную легенду

z = [6, 9, 13, 13, 16, 12, 20, 18, 17]
diff = [y[i] - z[i] for i in range(len(y))]

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(x, y, 'g', label='Ряд Y')
ax.plot(x, z, 'purple', label='Ряд Z')

ax2 = ax.twinx()
ax2.bar(x, diff, alpha=0.3, label='difference')
ax2.set_ylim(0, max(diff) + 2)

ax.legend(loc='lower center')
ax2.legend(loc='upper right')
plt.show()
```

### Задание 2. Поработаем с визуализацией датасета

```python
##############################################   Задание 2.1   #################################################################

##### Импортните csv-файл под названием Purchase_List
##### Приведите date_purchase к формату даты с помощью функции to_datetime
##### Создайте колонку датафрейма с месяцесем оплат (числом, от 1 до 12)
##### Сделайте группировку: количество оплат и сумма оплат для каждого месяца (создайте новый группировочный датафрейм)
##### Постройте визуализацию: для каждого месяца покупки по левой оси постройте сумму оплат, а по правой - количество оплат
##### Не забудьте оформление графика (заглавие, подпись осей, раздельная легенда, сетка)

import pandas as pd
import datetime as dt

df = pd.read_csv('Purchase_List.csv')
df['date_purchase'] = pd.to_datetime(df['date_purchase'])
df['month'] = df['date_purchase'].dt.month

grouped = df.groupby('month').agg(
    cnt=('id_purchase', 'count'),
    total=('amt_pay', 'sum')
).reset_index()

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(grouped['month'], grouped['total'], 'b', label='Сумма оплат')

ax2 = ax.twinx()
ax2.bar(grouped['month'], grouped['cnt'], alpha=0.3, color='green', label='Количество оплат')

ax.set_title('Оплаты по месяцам')
ax.set_xlabel('Месяц')
ax.set_ylabel('Сумма оплат')
ax2.set_ylabel('Количество оплат')
ax.grid()
ax.legend(loc='upper left')
ax2.legend(loc='upper right')
plt.show()
```

```python
##############################################   Задание 2.2   #################################################################

##### Постройте scatter plot для среднего чека для каждого месяца.
##### Для этого создайте новую колонку в сгруппированном датафрейме с отношением суммы платежей к их количеству
##### Сохраните форматирование из предыдущих пунктов
##### Задайте нижнюю границу оси Y как ноль. Верхнюю границу придумайте самостоятельно для лучшего отображения информации.

grouped['avg_check'] = grouped['total'] / grouped['cnt']

fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(grouped['month'], grouped['avg_check'], color='purple', s=100)

ax.set_title('Средний чек по месяцам')
ax.set_xlabel('Месяц')
ax.set_ylabel('Средний чек')
ax.set_ylim(0, grouped['avg_check'].max() + 500)
ax.grid()
plt.show()
```
