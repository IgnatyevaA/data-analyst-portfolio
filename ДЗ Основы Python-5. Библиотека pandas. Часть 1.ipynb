### Задание 1

Импортируйте файл workouts.csv.

Выведите тип полученного объекта и его первые 10 строк.

```python
import pandas as pd

workouts = pd.read_csv('workouts.csv')

print(type(workouts))
print(workouts.head(10))
```

### Задание 2

Воспроизведите результат следующего SQL-кода

~~~ sql
select   workout_id
       , start_at
       , client_id
from workouts
where  workout_schedule_type = 'trial'
limit 10
~~~

```python
result_2 = workouts[workouts['workout_schedule_type'] == 'trial'][['workout_id', 'start_at', 'client_id']].head(10)
print(result_2)
```

### Задание 3

Воспроизведите результат следующего SQL-кода

~~~ sql
select   workout_id
       , trainer_id
from workouts
where  workout_schedule_type in ('trial', 'regular')
  and trainer_department <> 'Sales'
order by trainer_id
limit 5
~~~

```python
result_3 = workouts[
    (workouts['workout_schedule_type'].isin(['trial', 'regular'])) & 
    (workouts['trainer_department'] != 'Sales')
].sort_values('trainer_id')[['workout_id', 'trainer_id']].head(5)
print(result_3)
```

### Задание 4

Воспроизведите результат следующего SQL-кода

~~~ sql
select   *
from workouts
where  trainer_department like '%Spartacus%'
   and workout_schedule_type not in ('trial', 'single')
order by client_id desc
limit 15
~~~

```python
result_4 = workouts[
    (workouts['trainer_department'].str.contains('Spartacus', na=False)) & 
    (~workouts['workout_schedule_type'].isin(['trial', 'single']))
].sort_values('client_id', ascending=False).head(15)
print(result_4)
```

### Задание 5

Воспроизведите результат следующего SQL-кода

~~~ sql
select   *
from workouts
where  status = 'success'
    or client_status = 'active' and workout_schedule_type = 'regular'
order by trainer_id
limit 5
~~~

```python
result_5 = workouts[
    (workouts['status'] == 'success') | 
    ((workouts['client_status'] == 'active') & (workouts['workout_schedule_type'] == 'regular'))
].sort_values('trainer_id').head(5)
print(result_5)
```

### Задание 6

Воспроизведите результат следующего SQL-кода

~~~ sql
select   workout_id
from workouts
where  client_status = 'new'
    and (trainer_department like 'Sales%' or trainer_department like '%Spartacus')
limit 10
~~~

```python
result_6 = workouts[
    (workouts['client_status'] == 'new') & 
    (
        (workouts['trainer_department'].str.startswith('Sales', na=False)) | 
        (workouts['trainer_department'].str.endswith('Spartacus', na=False))
    )
][['workout_id']].head(10)
print(result_6)
```

### Задание 7

Воспроизведите результат следующего SQL-кода с помощью предварительного создания двух масок.

~~~ sql
select   workout_id
from workouts
where  upper(trainer_department) like '%SPARTACUS%'
   and workout_schedule_type is not null
limit 10
~~~

Первая маска - upper(trainer_department) like '%SPARTACUS%'
<br>
Вторая маска - workout_schedule_type is not null

```python
mask_1 = workouts['trainer_department'].str.upper().str.contains('SPARTACUS', na=False)
mask_2 = workouts['workout_schedule_type'].notnull()

result_7 = workouts[mask_1 & mask_2][['workout_id']].head(10)
print(result_7)
```
